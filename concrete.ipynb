```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Load the dataset from the specified location
file_path = r"M:\Ai engineer\concrete_data.csv"  #  path to my CSV file
data = pd.read_csv(file_path)

# Explore the dataset
print("Dataset Preview:")
print(data.head())  # View the first few rows
print("\nDataset Information:")
print(data.info())  # Check for missing values and data types
```

```python
# Split predictors (X) and target (y)
X = data.drop("Strength", axis=1)  # Drop the target column
y = data["Strength"]
```

```python
# Placeholder for storing MSEs from 50 iterations
mse_list = []

# Repeat the experiment 50 times
for i in range(50):
    # Step 1: Split the data into training and test sets
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=i)
    
    # Step 2: Build the baseline model
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X_train.shape[1],)),  # Hidden layer with ReLU
        Dense(1)  # Output layer for regression
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')  # Compile with Adam optimizer and MSE loss
    
    # Step 3: Train the model on training data for 50 epochs
    model.fit(X_train, y_train, epochs=50, batch_size=10, verbose=1)  # Training with batch size of 10
    
```

```python
 # Step 4: Predict and evaluate the model
y_pred = model.predict(X_test)  # Predict on the test data
mse = mean_squared_error(y_test, y_pred)  # Compute Mean Squared Error
    
    # Append the MSE to the list
mse_list.append(mse)
```

```python
# Step 5: Compute mean and standard deviation of MSEs
mse_mean = np.mean(mse_list)
mse_std = np.std(mse_list)

# Print the results
print("\nResults After 50 Iterations:")
print(f"Mean MSE over 50 iterations: {mse_mean}")
print(f"Standard Deviation of MSEs: {mse_std}")
```

```python
# Normalize the predictors
X_normalized = (X - X.mean()) / X.std()  # Normalize each column
```

```python
# Placeholder for storing MSEs from 50 iterations for normalized data
mse_list_normalized = []

# Repeat the experiment 50 times
for i in range(50):
    # Step 1: Split the normalized data into training and test sets
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=i)
    
    # Step 2: Build the baseline model
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X_train.shape[1],)),  # Hidden layer with ReLU
        Dense(1)  # Output layer for regression
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')  # Compile with Adam optimizer and MSE loss
    
    # Step 3: Train the model on training data for 50 epochs
    model.fit(X_train, y_train, epochs=50, batch_size=10, verbose=0)  # Training with batch size of 10
    
```

```python
  # Step 4: Predict and evaluate the model
y_pred = model.predict(X_test)  # Predict on the test data
mse = mean_squared_error(y_test, y_pred)  # Compute Mean Squared Error
    
    # Append the MSE to the list
mse_list_normalized.append(mse)

# Step 5: Compute mean and standard deviation of MSEs for normalized data
mse_mean_normalized = np.mean(mse_list_normalized)
mse_std_normalized = np.std(mse_list_normalized)

# Print the results for Part B
print("\nResults After 50 Iterations (Part B - Normalized Data):")
print(f"Mean MSE over 50 iterations: {mse_mean_normalized}")
print(f"Standard Deviation of MSEs: {mse_std_normalized}")

# ----------------------------- Comparison Between Part A and Part B ----------------------------- #
print("\nComparison Between Part A and Part B:")
print(f"Mean MSE from Part A (Original Data): {mse_mean}")
print(f"Mean MSE from Part B (Normalized Data): {mse_mean_normalized}")
```

```python
# ----------------------------- Part D: Neural Network with Three Hidden Layers ----------------------------- #

# Placeholder for storing MSEs from 50 iterations
mse_list_three_hidden_layers = []

# Repeat the experiment 50 times
for i in range(50):
    # Step 1: Split the normalized data into training and test sets
    X_train, X_test, y_train, y_test = train_test_split(X_normalized, y, test_size=0.3, random_state=i)
    
    # Step 2: Build the model with three hidden layers
    model = Sequential([
        Dense(10, activation='relu', input_shape=(X_train.shape[1],)),  # First hidden layer with ReLU
        Dense(10, activation='relu'),                                  # Second hidden layer with ReLU
        Dense(10, activation='relu'),                                  # Third hidden layer with ReLU
        Dense(1)                                                       # Output layer for regression
    ])
    model.compile(optimizer='adam', loss='mean_squared_error')  # Compile with Adam optimizer and MSE loss
```

```python
 # Step 3: Train the model on training data for 50 epochs
    model.fit(X_train, y_train, epochs=50, batch_size=10, verbose=0)  # Training with batch size of 10
    
    # Step 4: Predict and evaluate the model
    y_pred = model.predict(X_test)  # Predict on the test data
    mse = mean_squared_error(y_test, y_pred)  # Compute Mean Squared Error
    
    # Append the MSE to the list
    mse_list_three_hidden_layers.append(mse)
```

```python
# Step 5: Compute mean and standard deviation of MSEs for three hidden layers
mse_mean_three_hidden_layers = np.mean(mse_list_three_hidden_layers)
mse_std_three_hidden_layers = np.std(mse_list_three_hidden_layers)
```

```python

# Print the results for Part D
print("\nResults After 50 Iterations (Part D - Three Hidden Layers):")
print(f"Mean MSE over 50 iterations: {mse_mean_three_hidden_layers}")
print(f"Standard Deviation of MSEs: {mse_std_three_hidden_layers}")
```
